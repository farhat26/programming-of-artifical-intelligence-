```python
import nltk
nltk.download('vader_lexicon')
nltk.download('punkt')
```

```python
from nltk.chat.util import Chat, reflections
from nltk.sentiment import SentimentIntensityAnalyzer
import re
```

```python
pairs = [
    [r"(hi|hello|hey)",
     ["Hello! Welcome to our Hospital ",
      "Hi there! How can I assist you today?"]],
    [r"(emergency|ambulance)",
     ["Emergency services are available 24/7 ",
      "Call 1122 for ambulance service immediately"]],
    [r"(departments|services)",
     ["We have Cardiology, Neurology, Orthopedics, Pediatrics, and General Medicine"]],
    [r"(doctors|doctor list)",
     ["Our top doctors include Dr. Ahmed (Cardiologist), Dr. Sara (Neurologist), Dr. Ali (Orthopedic)"]],
    [r"(visiting hours|timing)",
     ["Visiting hours are from 4 PM to 8 PM daily"]],
    [r"(room|availability|beds)",
     ["Rooms are available. Please contact reception for booking 🛏"]],
    [r"(appointment|book)",
     ["You can book an appointment by calling 042-1234567 "]],
    [r"(thanks|thank you)",
     ["You're welcome ", "Happy to help!"]],

    [r"(.*)",
     ["Sorry, I didn’t understand that. Can you rephrase?",
      "Please ask about hospital services, doctors, or emergency."]]
]
```

```python
chatbot = Chat(pairs, reflections)
sia = SentimentIntensityAnalyzer()
```

```python
def analyze_sentiment(text):
    score = sia.polarity_scores(text)

    if score['compound'] >= 0.05:
        return "positive"
    elif score['compound'] <= -0.05:
        return "negative"
    else:
        return "neutral"
```

```python
def chatbot_app():
    print("--- Hospital Information Chatbot ---")
    print("Type 'sentiment' to analyze your message")
    print("Type 'exit' to quit\n")

    while True:
        user_input = input("You: ").lower()

        if user_input == "exit":
            print("Bot: Thank you! Stay healthy ")
            break

        elif user_input == "sentiment":
            text = input("Enter text: ")
            result = analyze_sentiment(text)
            print("Sentiment:", result)

        else:
            response = chatbot.respond(user_input)
            print("Bot:", response)
```

```python
chatbot_app()
```
