```python
capa = 4
capb = 3
goal = 2

a = 0
b = 0

visited = []

path = []
path.append(f"Start State: ({a}, {b})")

stack = [(a, b, path)]

found = False

while stack:
    a, b, path = stack.pop()

    if a == goal or b == goal:
        print("Goal reached!")
        for step in path:
            print(step)
        found = True
        break

    if (a, b) in visited:
        continue
    visited.append((a, b))

    if a < capa and (capa, b) not in visited:
        newpath = path + [f"Fill Jug A  : ({capa}, {b})"]
        stack.append((capa, b, newpath))

    if b < capb and (a, capb) not in visited:
        newpath = path + [f"Fill Jug B : ({a}, {capb})"]
        stack.append((a, capb, newpath))

    if a > 0 and (0, b) not in visited:
        newpath = path + [f"Empty Jug A  : (0, {b})"]
        stack.append((0, b, newpath))

    if b > 0 and (a, 0) not in visited:
        newpath = path + [f"Empty Jug B  : ({a}, 0)"]
        stack.append((a, 0, newpath))

    if a > 0 and b < capb:
        pour = min(a, capb - b)
        new_a = a - pour
        new_b = b + pour
        if (new_a, new_b) not in visited:
            newpath = path + [f"Pour {pour}L from A to B : ({new_a}, {new_b})"]
            stack.append((new_a, new_b, newpath))

    if b > 0 and a < capa:
        pour = min(b, capa - a)
        new_a = a + pour
        new_b = b - pour
        if (new_a, new_b) not in visited:
            newpath = path + [f"Pour {pour}L from B to A : ({new_a}, {new_b})"]
            stack.append((new_a, new_b, newpath))

if not found:
    print("No solution found")
```
