```python
n = 4
values = [[' ' for i in range(n)] for i in range(n)]
queens = []  

horizontal_line = "+" + "-----+" * n

for i in range(n):
    print(horizontal_line)
    for j in range(n):
        print(f"|  {values[i][j]}  ", end="")
    print("|")
print(horizontal_line)

while True:
    if len(queens) == n:
        print("All queens placed successfully! You won!")
        break

    user_input = input("Enter queen position as row,col (or 'exit' to quit): ")
    if user_input.lower() == 'exit':
        break

    try:
        row, col = map(int, user_input.split(','))
        if row < 0 or row >= n or col < 0 or col >= n:
            print("Invalid position. Try again.")
            continue
    except:
        print("Invalid input. Use row,col format.")
        continue

    safe = True
    for qrow, qcol in queens:
        if qrow == row or qcol == col or abs(qrow - row) == abs(qcol - col):
            safe = False
            break

    if safe:
        if values[row][col] == 'Q':
            print("Queen already placed here!")
            continue
        values[row][col] = 'Q'
        queens.append((row, col))
        print("Queen placed successfully!")
    else:
        print("Cannot place queen here! Conflicts with existing queens.")

    for i in range(n):
        print(horizontal_line)
        for j in range(n):
            print(f"|  {values[i][j]}  ", end="")
        print("|")
    print(horizontal_line)
```
